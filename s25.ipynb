## Q1. Bar Chart of Course Enrolments

```python
import matplotlib.pyplot as plt

courses = ["BCA", "BBA", "B.Com", "B.Sc", "BA"]
students = [60, 45, 75, 50, 35]
plt.figure(figsize=(7, 5))
plt.bar(courses, students, color=["skyblue", "orange", "green", "pink", "purple"])
plt.xlabel("Course")
plt.ylabel("Number of Students")
plt.title("Students Enrolled in Different Courses")
plt.tight_layout()
plt.show()
```

## Q2. Mobile Data Integration and Encoding

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

phones1 = pd.DataFrame({
    "Brand": ["Samsung", "Apple", "OnePlus", "Samsung"],
    "Model": ["A15", "iPhone 14", "Nord", "S24"],
    "Price": [18000, 65000, 30000, 75000],
    "RAM": [6, 6, 8, 8]
})
phones2 = pd.DataFrame({
    "Model": ["A15", "iPhone 14", "Nord", "S24"],
    "Storage": [128, 128, 256, 256],
    "Battery_Capacity": [5000, 3279, 5000, 4000],
    "Operating_System": ["Android", "iOS", "Android", "Android"]
})
print("DataFrame 1:")
print(phones1)
print("\nDataFrame 2:")
print(phones2)
merged = pd.merge(phones1, phones2, on="Model", how="inner")
print("\nMerged DataFrame:")
print(merged)
merged["Brand_Encoded"] = LabelEncoder().fit_transform(merged["Brand"])
result = pd.get_dummies(merged, columns=["Operating_System"], dtype=int)
print("\nTransformed DataFrame:")
print(result)
```

## Q2 OR. Groceries Apriori, support 0.3 and 0.4

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

transactions = [
    ["Basmati Rice", "Toor Dal", "Turmeric Powder", "Cooking Oil"],
    ["Atta", "Sugar", "Tea", "Milk"],
    ["Rice", "Moong Dal", "Salt", "Cooking Oil"],
    ["Wheat Flour", "Chickpeas", "Turmeric Powder", "Cumin Seeds"],
    ["Poha", "Peanuts", "Onions", "Cooking Oil"],
    ["Basmati Rice", "Paneer", "Garam Masala", "Tomatoes"],
    ["Atta", "Potatoes", "Onions", "Cooking Oil"]
]
encoder = TransactionEncoder()
encoded = encoder.fit(transactions).transform(transactions)
basket = pd.DataFrame(encoded, columns=encoder.columns_)
for support in [0.3, 0.4]:
    print("\nMinimum support:", support)
    frequent = apriori(basket, min_support=support, use_colnames=True)
    print(frequent)
    if not frequent.empty:
        rules = association_rules(frequent, metric="confidence", min_threshold=0.5)
        print(rules[["antecedents", "consequents", "support", "confidence", "lift"]])
```
