## Q1. Weather Parameters Correlation Heat Map

> **Dataset:** `weather.csv` — this cell creates the CSV automatically (sample data) if it is not present. If you have the real file in the same folder, the code uses that instead.

```python
# ---- Sample dataset: CSV is created automatically if it is not in the folder ----
import os
import numpy as np
import pandas as pd
if not os.path.exists('weather.csv'):
    rng = np.random.default_rng(42)
    n = 200
    temp = np.round(rng.uniform(15, 40, n), 1)
    hum = (95 - 1.5 * (temp - 15) + rng.normal(0, 8, n)).clip(10, 100).round(1)
    pd.DataFrame({"Date": pd.date_range("2025-01-01", periods=n).astype(str),
        "Temperature": temp, "Humidity": hum,
        "Rainfall": (np.maximum(0, hum - 55) * rng.uniform(0.1, 0.6, n)).round(1),
        "Wind Speed": np.round(rng.uniform(2, 30, n), 1)}).to_csv('weather.csv', index=False)
    print('weather.csv', 'created')
# ---- Your program starts here ----

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("weather.csv")
print(df.head())
# Update these names if your CSV uses different column labels.
columns = ["Temperature", "Humidity", "Rainfall", "Wind Speed"]
df = df[columns].dropna()
correlation = df.corr()
print(correlation)
sns.heatmap(correlation, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Heat Map of Weather Parameters")
plt.tight_layout()
plt.show()
```

## Q2. Student K-Means Clustering, K=3

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
    "Student": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8"],
    "Attendance": [55, 60, 65, 72, 78, 85, 90, 95],
    "Marks": [40, 45, 50, 62, 70, 78, 88, 94]
})
X = df[["Attendance", "Marks"]]
X_scaled = StandardScaler().fit_transform(X)
model = KMeans(n_clusters=3, random_state=42, n_init=10)
df["Cluster"] = model.fit_predict(X_scaled)
print(df)
centers = StandardScaler().fit(X).inverse_transform(model.cluster_centers_)
plt.scatter(df["Attendance"], df["Marks"], c=df["Cluster"], cmap="viridis", s=70)
plt.scatter(centers[:, 0], centers[:, 1], c="red", marker="X", s=200, label="Centroids")
plt.xlabel("Attendance")
plt.ylabel("Marks")
plt.title("Student Clusters (K=3)")
plt.legend()
plt.tight_layout()
plt.show()
```

## Q2 OR. Loan Application Feature Selection and PCA

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.decomposition import PCA

data = {
    "Application_ID": range(1, 13),
    "Age": [25, 42, 31, 50, 28, 39, 45, 23, 36, 48, 29, 41],
    "Income": [30000, 70000, 42000, 90000, 35000, 65000, 80000, 28000, 50000, 85000, 39000, 72000],
    "Credit_Score": [620, 760, 680, 810, 640, 740, 790, 600, 700, 800, 660, 750],
    "Loan_Amount": [8000, 20000, 10000, 25000, 9000, 18000, 22000, 7000, 13000, 24000, 9500, 19000],
    "Employment_Years": [2, 12, 5, 18, 3, 10, 15, 1, 7, 17, 4, 11],
    "Debt_Ratio": [0.45, 0.20, 0.35, 0.15, 0.40, 0.22, 0.18, 0.50, 0.30, 0.16, 0.38, 0.21],
    "Loan_Status": ["Rejected", "Approved", "Rejected", "Approved", "Rejected", "Approved", "Approved", "Rejected", "Approved", "Approved", "Rejected", "Approved"]
}
df = pd.DataFrame(data)
print("Original data:")
print(df)
X = df[["Age", "Income", "Credit_Score", "Loan_Amount", "Employment_Years", "Debt_Ratio"]]
y = df["Loan_Status"].map({"Rejected": 0, "Approved": 1})
selector = SelectKBest(score_func=f_classif, k=3)
X_selected = selector.fit_transform(X, y)
print("\nSelected features:")
print(X.columns[selector.get_support()].tolist())
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
print("\nPCA reduced data (2 components):")
print(pd.DataFrame(X_pca, columns=["PC1", "PC2"]).round(3))
print("Explained variance ratio:", pca.explained_variance_ratio_)
```
