## Q1. Student Performance Correlation Heat Map

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.DataFrame({
    "Study_Hours": [2, 3, 4, 5, 6, 7],
    "Attendance": [60, 65, 70, 80, 85, 95],
    "Assignment_Score": [50, 55, 65, 70, 80, 90],
    "Exam_Score": [45, 50, 60, 72, 82, 94]
})
correlation = df.corr()
print(correlation)
sns.heatmap(correlation, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Heat Map of Student Performance")
plt.tight_layout()
plt.show()
```

## Q2. Min-Max Scaling, Standardization and Normalization

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, Normalizer

df = pd.DataFrame({
    "Temperature": [20, 25, 30, 35, 40],
    "Humidity": [40, 50, 60, 70, 80],
    "Wind-Speed": [5, 8, 10, 12, 15],
    "Pressure": [1000, 1005, 1010, 1015, 1020],
    "Rainfall": [0, 5, 10, 15, 20]
})
print("Original Data:")
print(df)
columns = df.columns
minmax = pd.DataFrame(MinMaxScaler().fit_transform(df), columns=columns)
standard = pd.DataFrame(StandardScaler().fit_transform(df), columns=columns)
normalized = pd.DataFrame(Normalizer().fit_transform(df), columns=columns)
print("\nMin-Max Scaling:")
print(minmax.round(3))
print("\nStandardization:")
print(standard.round(3))
print("\nNormalization:")
print(normalized.round(3))
```

## Q2 OR. Student Activity Apriori, support 0.2 and 0.3

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

transactions = [
    ["Library", "Seminar"], ["Workshop", "Project"],
    ["Library", "Assignment"], ["Seminar", "Workshop"],
    ["Assignment", "Seminar"], ["Workshop", "Assignment"],
    ["Project", "Library"], ["Library", "Seminar", "Project"],
    ["Library", "Seminar", "Workshop"]
]
encoder = TransactionEncoder()
basket = pd.DataFrame(encoder.fit(transactions).transform(transactions), columns=encoder.columns_)
for support in [0.2, 0.3]:
    print("\nMinimum support:", support)
    frequent = apriori(basket, min_support=support, use_colnames=True)
    print(frequent)
    if not frequent.empty:
        rules = association_rules(frequent, metric="confidence", min_threshold=0.4)
        print(rules[["antecedents", "consequents", "support", "confidence", "lift"]])
```
