## Q1. House Rent Data Normalization

```python
import pandas as pd
from sklearn.preprocessing import Normalizer

df = pd.DataFrame({
    "City": ["Pune", "Mumbai", "Nashik", "Nagpur", "Thane"],
    "House_Type": ["Flat", "Apartment", "House", "Flat", "Apartment"],
    "House_Area": [600, 900, 750, 850, 700],
    "Number_of_Rooms": [1, 2, 2, 3, 2],
    "Monthly_Rent": [12000, 25000, 14000, 16000, 22000]
})
print("Original Data:")
print(df)
numeric_columns = ["House_Area", "Number_of_Rooms", "Monthly_Rent"]
normalized_values = Normalizer().fit_transform(df[numeric_columns])
normalized = df.copy()
normalized[numeric_columns] = normalized_values
print("\nNormalized Data:")
print(normalized.round(3))
```

## Q2. Netflix and Amazon Prime Venn Diagram and Pie Chart

```python
import matplotlib.pyplot as plt
from matplotlib_venn import venn2
import pandas as pd

customers = pd.DataFrame({
    "Customer": ["A", "B", "C", "D", "E", "F", "G"],
    "Netflix": [1, 1, 1, 0, 0, 1, 0],
    "Amazon_Prime": [0, 1, 0, 1, 0, 1, 1]
})
netflix = set(customers.loc[customers["Netflix"] == 1, "Customer"])
prime = set(customers.loc[customers["Amazon_Prime"] == 1, "Customer"])
only_netflix = len(netflix - prime)
only_prime = len(prime - netflix)
both = len(netflix & prime)
plt.figure(figsize=(6, 5))
venn2(subsets=(only_netflix, only_prime, both), set_labels=("Netflix", "Amazon Prime"))
plt.title("Netflix and Amazon Prime Users")
plt.tight_layout()
plt.show()
labels = ["Only Netflix", "Only Amazon Prime", "Both"]
values = [only_netflix, only_prime, both]
plt.figure(figsize=(6, 5))
plt.pie(values, labels=labels, autopct="%1.1f%%", startangle=90)
plt.title("Streaming Service Customer Distribution")
plt.tight_layout()
plt.show()
```

## Q2 OR. Student K-Means Clustering, K=3

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
    "Student": ["S1", "S2", "S3", "S4", "S5", "S6", "S7", "S8"],
    "Attendance": [55, 60, 65, 72, 78, 85, 90, 95],
    "Marks": [40, 45, 50, 62, 70, 78, 88, 94]
})
X = df[["Attendance", "Marks"]]
X_scaled = StandardScaler().fit_transform(X)
model = KMeans(n_clusters=3, random_state=42, n_init=10)
df["Cluster"] = model.fit_predict(X_scaled)
print(df)
centers = StandardScaler().fit(X).inverse_transform(model.cluster_centers_)
plt.scatter(df["Attendance"], df["Marks"], c=df["Cluster"], cmap="viridis", s=70)
plt.scatter(centers[:, 0], centers[:, 1], c="red", marker="X", s=200, label="Centroids")
plt.xlabel("Attendance")
plt.ylabel("Marks")
plt.title("Student Clusters (K=3)")
plt.legend()
plt.tight_layout()
plt.show()
```
