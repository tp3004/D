## Q1. Standardization of Age, Salary, Experience and other attributes

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
    "Age": [22, 25, 30, 35, 40],
    "Salary": [18000, 25000, 35000, 45000, 60000],
    "Experience": [1, 3, 5, 8, 12],
    "Working_Hours": [6, 7, 8, 8, 9],
    "Performance_Score": [60, 65, 72, 80, 90]
})
print("Original Data:")
print(df)
scaled = pd.DataFrame(StandardScaler().fit_transform(df), columns=df.columns)
print("\nAfter Standardization:")
print(scaled.round(3))
```

## Q2. Books Venn Diagram and Word Cloud

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib_venn import venn2
from wordcloud import WordCloud

df = pd.DataFrame({
    "Book Title": ["Data Science Basics", "Python for Everyone", "Machine Learning Guide",
                   "Advanced Analytics", "Learning Python", "AI and Data"],
    "Author": ["Asha", "Ravi", "Asha", "Meera", "Ravi", "John"],
    "Publication Year": [2005, 2010, 2018, 2021, 2008, 2022],
    "Average Rating": [4.1, 4.3, 4.6, 4.0, 4.2, 4.5],
    "Number of Ratings": [100, 150, 200, 90, 120, 170]
})
print(df)
older_authors = set(df.loc[df["Publication Year"] < 2015, "Author"])
newer_authors = set(df.loc[df["Publication Year"] >= 2015, "Author"])
plt.figure(figsize=(6, 5))
venn2([older_authors, newer_authors], set_labels=("Before 2015", "2015 and later"))
plt.title("Authors by Publication Period")
plt.tight_layout()
plt.show()
text = " ".join(df["Book Title"].astype(str))
cloud = WordCloud(width=800, height=400, background_color="white").generate(text)
plt.figure(figsize=(10, 5))
plt.imshow(cloud, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud of Book Titles")
plt.tight_layout()
plt.show()
```

## Q2 OR. Student Activity Apriori, support 0.2 and 0.3

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

transactions = [
    ["Library", "Seminar"], ["Workshop", "Project"],
    ["Library", "Assignment"], ["Seminar", "Workshop"],
    ["Assignment", "Seminar"], ["Workshop", "Assignment"],
    ["Project", "Library"], ["Library", "Seminar", "Project"],
    ["Library", "Seminar", "Workshop"]
]
encoder = TransactionEncoder()
basket = pd.DataFrame(encoder.fit(transactions).transform(transactions), columns=encoder.columns_)
for support in [0.2, 0.3]:
    print("\nMinimum support:", support)
    frequent = apriori(basket, min_support=support, use_colnames=True)
    print(frequent)
    if not frequent.empty:
        rules = association_rules(frequent, metric="confidence", min_threshold=0.4)
        print(rules[["antecedents", "consequents", "support", "confidence", "lift"]])
```
