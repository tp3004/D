## Q1. Product DataFrame and Dendrogram

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage

data = {
    "Product": ["P1", "P2", "P3", "P4", "P5"],
    "Price": [100, 120, 500, 520, 300],
    "Rating": [4.1, 4.0, 4.8, 4.7, 3.9],
    "Number_of_Sales": [50, 55, 90, 85, 60]
}
df = pd.DataFrame(data)
print(df)
X = df[["Price", "Rating", "Number_of_Sales"]]
Z = linkage(X, method="ward")
plt.figure(figsize=(8, 5))
dendrogram(Z, labels=df["Product"].tolist())
plt.title("Dendrogram of Similar Products")
plt.xlabel("Product")
plt.ylabel("Distance")
plt.tight_layout()
plt.show()
```

## Q2. Loan Application Feature Selection and PCA

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.decomposition import PCA

data = {
    "Application_ID": range(1, 13),
    "Age": [25, 42, 31, 50, 28, 39, 45, 23, 36, 48, 29, 41],
    "Income": [30000, 70000, 42000, 90000, 35000, 65000, 80000, 28000, 50000, 85000, 39000, 72000],
    "Credit_Score": [620, 760, 680, 810, 640, 740, 790, 600, 700, 800, 660, 750],
    "Loan_Amount": [8000, 20000, 10000, 25000, 9000, 18000, 22000, 7000, 13000, 24000, 9500, 19000],
    "Employment_Years": [2, 12, 5, 18, 3, 10, 15, 1, 7, 17, 4, 11],
    "Debt_Ratio": [0.45, 0.20, 0.35, 0.15, 0.40, 0.22, 0.18, 0.50, 0.30, 0.16, 0.38, 0.21],
    "Loan_Status": ["Rejected", "Approved", "Rejected", "Approved", "Rejected", "Approved", "Approved", "Rejected", "Approved", "Approved", "Rejected", "Approved"]
}
df = pd.DataFrame(data)
print("Original data:")
print(df)
X = df[["Age", "Income", "Credit_Score", "Loan_Amount", "Employment_Years", "Debt_Ratio"]]
y = df["Loan_Status"].map({"Rejected": 0, "Approved": 1})
selector = SelectKBest(score_func=f_classif, k=3)
X_selected = selector.fit_transform(X, y)
print("\nSelected features:")
print(X.columns[selector.get_support()].tolist())
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
print("\nPCA reduced data (2 components):")
print(pd.DataFrame(X_pca, columns=["PC1", "PC2"]).round(3))
print("Explained variance ratio:", pca.explained_variance_ratio_)
```

## Q2 OR. K-Means on Iris dataset, K=3

```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans

iris = load_iris()
X = iris.data[:, [0, 1]]
model = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = model.fit_predict(X)
centers = model.cluster_centers_
plt.figure(figsize=(7, 5))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap="viridis", s=40)
plt.scatter(centers[:, 0], centers[:, 1], c="red", marker="X", s=200, label="Centroids")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("K-Means Clustering on Iris Dataset (K=3)")
plt.legend()
plt.tight_layout()
plt.show()
print("Cluster centroids:")
print(centers)
```
